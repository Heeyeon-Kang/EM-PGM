```python
### PGM_diabetes_diagnosis ###
PGM_optimal_k = PGM_diabetes_diagnosis["k"]

## correlation ##
PGM_corr = []
for i in range(PGM_optimal_k):
    PGM_corr.append(PGM_diabetes_diagnosis["optimal"]["sigma"][i][0,1] / np.sqrt(PGM_diabetes_diagnosis["optimal"]["sigma"][i][0,0] * PGM_diabetes_diagnosis["optimal"]["sigma"][i][1,1]))

## coefficients ##
intercept = pd.DataFrame({"intercept": [1]})
gender = pd.DataFrame({"gender": [1]})
sd_X = pd.concat([intercept, pd.DataFrame(dat.diabetes_X1.std()).T.reset_index(drop=True), gender, pd.DataFrame(dat.diabetes_X2.std()).T.reset_index(drop=True)], axis=1)
PGM_new_Bk = [np.zeros_like(x) for x in PGM_diabetes_diagnosis["optimal"]["Bk"]]
sd = sd_X.iloc[0].to_numpy()
for k in range(PGM_optimal_k):
    PGM_new_Bk[k] = PGM_diabetes_diagnosis["optimal"]["Bk"][k] / sd[:, None]
```
